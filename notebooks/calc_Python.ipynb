```python
g = Graph()

g.insert_node("/data/a/a1/prop_a", "")
g.insert_node("/data/a/a1/prop_b", "")
g.insert_node("/data/b/b1/prop_a", "")
g.insert_node("/data/b/b1/prop_b", "")
g.insert_node("/data/b/b2/prop_a", "")
g.insert_node("/data/b/b2/prop_b", "")
g.register_function(
    name="f",
    template="{a}",
    forEach=[[["table", "key"], "/data/*"]],
    context={"a": "/data/{table}/{key}/prop_a"},
    target="/data/{table}/{key}/prop_b"
)
```

```python
g.register_change("/data/b/b2/prop_a", "z")
for k, v in g.nodes.items():
    if v["type"]=="data":
        print(k, v["value"])
```
